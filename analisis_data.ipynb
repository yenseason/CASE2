```python
import pandas as pd
import numpy as np

np.random.seed(42)

jurusan_list = ['Sistem Informasi', 'Manajemen', 'Teknik Informatika', 'Akuntansi', 'Desain Komunikasi Visual']
n_samples = 1000

data = []
for _ in range(n_samples):
    jurusan = np.random.choice(jurusan_list, p=[0.25, 0.25, 0.20, 0.15, 0.15])
    semester = np.random.randint(1, 9)
    usia = 17 + int(semester * 0.7) + np.random.choice([0, 1])
    ipk = round(np.random.normal(loc=3.35, scale=0.35), 2)
    ipk = min(4.00, max(2.00, ipk))

    data.append({
        'Jurusan': jurusan,
        'Semester': semester,
        'IPK': ipk,
        'Usia': usia
    })

df = pd.DataFrame(data)
df.to_csv('dataset_mahasiswa.csv', index=False)
print("✅ SUCCESS: 1.000 Data Sintetis Mahasiswa Berhasil Dibuat!")
```
